# Confirmatory hypothesis tests (VS)

Preregistered H1-H4 tests for the confirmatory VS design. Each criterion is
tested as a two-sided paired $t$ test on the 90 per-seed condition
differences, with the unadjusted 95% $t$ interval as uncertainty and
Benjamini-Hochberg as the multiple-testing correction (four tasks x eight
directional tests per model). A hypothesis is supported when all of its
criteria are met.
